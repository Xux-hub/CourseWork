```python
## add your code here
import java.io.IOException;
import java.io.InputStream;
import java.util.ArrayList;
import java.util.Collections;

public class Main {
    static class Reader {
        private InputStream in;
        private byte[] buf = new byte[131072];
        private int head = 0;
        private int tail = 0;

        public Reader(InputStream in) {
            this.in = in;
        }

        public int read() throws IOException {
            if (head >= tail) {
                head = 0;
                tail = in.read(buf, 0, buf.length);
                if (tail <= 0) return -1;
            }
            return buf[head++];
        }

        public int nextInt() throws IOException {
            int c = read();
            while (c < '0') {
                if (c == -1) return -1;
                c = read();
            }
            int res = 0;
            while (c >= '0') {
                res = res * 10 + (c - '0');
                c = read();
            }
            return res;
        }
    }

    static final int N = 1024;
    static ArrayList<Integer> ans = new ArrayList<>();
    static int[] o = new int[N];
    static int[] posi = new int[N];
    static int a, b, l, n;

    static void rf() {
        for (int i = 0; i < n; i++) {
            posi[o[i]] = i;
        }
    }

    static void doMagic() {
        ans.add(0);
        for (int i = 0; i < n; i++) {
            if (o[i] == a) o[i] = b;
            else if (o[i] == b) o[i] = a;
        }
        rf();
    }

    static void doAdd(int x) {
        if (x == 0) return;
        ans.add(x);
        for (int i = 0; i < n; i++) {
            o[i] = (o[i] + x) % n;
        }
        rf();
    }

    static void doXor(int x) {
        if (x == 0) return;
        ans.add(-x);
        for (int i = 0; i < n; i++) {
            o[i] = (o[i] ^ x);
        }
        rf();
    }

    static int[] calc(int a_val, int b_val) {
        int delta = (b_val - a_val + n - l + n) % n;
        int pa = 0, pb = 0;
        for (int stp = n / 2; stp >= 2 * l; stp /= 2) {
            if (delta >= stp) {
                delta -= stp;
                pb += stp / 2;
            } else {
                pa += stp / 2;
            }
        }
        pa += n / 2;
        pa += (a_val & (l - 1));
        pb += (a_val & (l - 1));
        return new int[]{pa, pb};
    }

    static void doSwap(int c, int d) {
        if ((c / l % 2) == (d / l % 2)) {
            int p;
            if ((c / l % 2) == 0) {
                p = (c & (l - 1)) + l;
            } else {
                p = (c & (l - 1));
            }
            doSwap(c, p);
            doSwap(d, p);
            doSwap(c, p);
        } else {
            int[] p_ab = calc(a, b);
            int pa = p_ab[0], pb = p_ab[1];
            
            int[] p_cd = calc(c, d);
            int pc = p_cd[0], pd = p_cd[1];

            doAdd((pc - c + n) % n);
            doXor((pc ^ pa));
            doAdd((a - pa + n) % n);
            doMagic();
            doAdd((pa - a + n) % n);
            doXor((pc ^ pa));
            doAdd((c - pc + n) % n);
        }
    }

    static class Perm {
        int[] arr = new int[N];
        int sz;
        ArrayList<Integer> vec = new ArrayList<>();

        boolean calc() {
            if (sz == 1) return true;
            
            Perm p_b = new Perm();
            Perm p_c = new Perm();
            p_b.sz = sz / 2;
            p_c.sz = sz / 2;
            
            for (int i = 0; i < sz / 2; i++) {
                p_b.arr[i] = arr[i * 2] / 2;
                p_c.arr[i] = arr[i * 2 + 1] / 2;
            }
            if (!p_b.calc() || !p_c.calc()) return false;
            
            if (arr[0] % 2 != 0) {
                vec.add(sz == 2 ? 1 : -1);
            }
            
            int tb = 0;
            for (int i : p_b.vec) {
                if (i > 0) {
                    vec.add(-1);
                    vec.add(1);
                } else {
                    vec.add(i * 2);
                    tb ^= -i * 2;
                }
            }
            if (tb != 0) vec.add(-tb);
            
            int tc = 0;
            for (int i : p_c.vec) {
                if (i > 0) {
                    vec.add(1);
                    vec.add(-1);
                } else {
                    vec.add(i * 2);
                    tc ^= -i * 2;
                }
            }
            
            if ((tc & (sz / 2)) != (tb & (sz / 2))) {
                for (int i = 0; i < sz / 4; i++) {
                    vec.add(-1);
                    vec.add(1);
                }
            }
            
            if (tb >= (sz / 2)) tb -= sz / 2;
            if (tc >= (sz / 2)) tc -= sz / 2;
            if (tb != tc) return false;
            
            ArrayList<Integer> tmp = new ArrayList<>();
            for (int i : vec) {
                if (tmp.isEmpty()) {
                    tmp.add(i);
                } else {
                    int back = tmp.get(tmp.size() - 1);
                    if (i < 0 && back < 0) {
                        int newVal = -((-back) ^ (-i));
                        tmp.set(tmp.size() - 1, newVal);
                        if (tmp.get(tmp.size() - 1) == 0) {
                            tmp.remove(tmp.size() - 1);
                        }
                    } else {
                        tmp.add(i);
                    }
                }
            }
            vec = tmp;
            return true;
        }
    }

    public static void main(String[] args) throws IOException {
        Reader in = new Reader(System.in);
        
        n = in.nextInt();
        a = in.nextInt();
        b = in.nextInt();
        for (int i = 0; i < n; i++) {
            o[i] = in.nextInt();
        }

        rf();
        l = (a - b + n) % n;
        l = l & -l;
        if (l == 0) l = n;
        
        if (l > 1) {
            Perm p_a = new Perm();
            p_a.sz = l;
            for (int i = 0; i < n; i++) {
                p_a.arr[i] = o[i] & (l - 1);
            }
            if (!p_a.calc()) {
                System.out.println("-1");
                System.exit(0);
            }
            for (int i : p_a.vec) {
                if (i > 0) doAdd(i);
                else doXor(-i);
            }
        }

        for (int i = 0; i < l; i++) {
            ArrayList<Integer> vec = new ArrayList<>();
            for (int j = i; j < n; j += l) {
                vec.add(o[j]);
            }
            
            int c = 0, flag = 1;
            Collections.sort(vec);
            
            for (int j = i; j < n; j += l) {
                if (vec.get(c) != j) {
                    flag = 0;
                    break;
                }
                c++;
            }
            
            if (flag == 0) {
                System.out.println("-1");
                System.exit(0);
            }
            
            for (int j = i; j < n; j += l) {
                if (o[j] != j) {
                    doSwap(j, o[j]);
                }
            }
        }
        
        for (int i = 0; i < n; i++) {
            if (o[i] != i) {
                throw new RuntimeException("Assertion failed: o[" + i + "] != " + i);
            }
        }
        
        StringBuilder sb = new StringBuilder();
        sb.append(ans.size()).append("\n");
        for (int i : ans) {
            if (i == 0) {
                sb.append("0\n");
            } else if (i < 0) {
                sb.append("1 ").append(-i).append("\n");
            } else {
                sb.append("2 ").append(i).append("\n");
            }
        }
        System.out.print(sb.toString());
    }
}
```

```python
## add your code here
import java.io.*;

public class Main {
    static BufferedReader br = new BufferedReader(new InputStreamReader(System.in));
    static StreamTokenizer st = new StreamTokenizer(br);
    
    public static void main(String[] args) throws IOException { 
        solveProblem(); 
    }

    public static String ins() throws IOException {
        st.nextToken();
        return st.sval;
    }

    public static int ini() throws IOException {
        st.nextToken();
        return (int) st.nval;
    }

    public static void solveProblem() throws IOException {
        int n = ini();
        String s1 = ins();
        String s2 = ins();

        char[] str1 = getArray(s1);
        char[] str2 = getArray(s2);
        
        int len = str1.length;
        int[] radius1 = new int[len];
        int[] radius2 = new int[len];
        
        manacher(str1, radius1);
        manacher(str2, radius2);

        int maxLen = 0;

        for (int i = 2; i <= len - 1; i++) {
            int t = Math.max(radius1[i], radius2[i - 2]);
            
            while (i - t >= 0 && i + t <= len - 1 && str1[i - t] == str2[i + t - 2]) {
                t++;
            }
            if (t > maxLen) {
                maxLen = t;
            }
        }

        System.out.println(maxLen - 1);
    }

    public static char[] getArray(String s) {
        char[] res = new char[s.length() * 2 + 2];
        res[0] = '@';
        int idx = 1;
        for (int i = 0; i < s.length(); i++) {
            res[idx++] = '#';
            res[idx++] = s.charAt(i);
        }
        res[idx] = '#';
        return res;
    }

    public static void manacher(char[] s, int[] radius) {
        int r = -1, c = -1;
        int len = s.length - 1;
        for (int i = 0; i <= len; i++) {
            radius[i] = i < r ? Math.min(radius[c * 2 - i], r - i + 1) : 1;
            while (i + radius[i] <= len && i - radius[i] > -1 && s[i + radius[i]] == s[i - radius[i]]) {
                radius[i]++;
            }
            if (i + radius[i] > r) {
                r = i + radius[i] - 1;
                c = i;
            }
        }
    }
}
```

```python
## add your code here
import java.io.*;
import java.util.*;

public class Main {
    
    static class FastScanner {
        BufferedReader br;
        StringTokenizer st;

        public FastScanner() {
            br = new BufferedReader(new InputStreamReader(System.in));
        }

        String next() {
            while (st == null || !st.hasMoreElements()) {
                try {
                    String line = br.readLine();
                    if (line == null) return null;
                    st = new StringTokenizer(line);
                } catch (IOException e) {
                    e.printStackTrace();
                }
            }
            return st.nextToken();
        }

        int nextInt() {
            return Integer.parseInt(next());
        }
    }

    public static void main(String[] args) {
        FastScanner scan = new FastScanner();
        String firstToken;
        
        while ((firstToken = scan.next()) != null) {
            int m = Integer.parseInt(firstToken);
            int ans = -1;
            boolean flag = false;
            
            TreeSet<Integer> ques = new TreeSet<>();
            Map<Integer, Integer> lastLine = new HashMap<>();
            
            Map<Integer, Character> lastOp = new HashMap<>();
            
            for (int i = 1; i <= m; i++) {
                String opStr = scan.next();
                char op = opStr.charAt(0);
                int id = 0;
                
                if (op == 'O' || op == 'I') {
                    id = scan.nextInt();
                }
                
                if (flag) {
                    continue;
                }
                
                if (op == '?') {
                    ques.add(i);
                    continue;
                }
                
                if (lastLine.containsKey(id)) {
                    int prevLine = lastLine.get(id);
                    if (lastOp.get(id) == op) {
                        Integer queLine = ques.higher(prevLine);
                        if (queLine == null) {
                            ans = i;
                            flag = true;
                        } else {
                            ques.remove(queLine);
                        }
                    }
                } else {
                    if (op == 'O') {
                        Integer queLine = ques.higher(0);
                        if (queLine == null) {
                            ans = i;
                            flag = true;
                        } else {
                            ques.remove(queLine);
                        }
                    }
                }
                
                lastLine.put(id, i);
                lastOp.put(id, op);
            }
            
            System.out.println(ans);
        }
    }
}
```

## F 通配符匹配

```python
## add your code here
import java.io.*;

public class Main {
    static final int N = 100005;
    static final long M = 1000000007L;
    
    static int[] p = new int[20]; 
    static long[] hs = new long[N];
    static long[] ht = new long[N];
    static long[] g = new long[N];
    static boolean[][] f = new boolean[20][N];
    static char[] s = new char[N];
    static char[] t = new char[N];

    static long hashs(int l, int r) {
        if (l > r) return 0; 
        return hs[r] - hs[l - 1] * g[r - l + 1];
    }

    static long hasht(int l, int r) {
        if (l > r) return 0;
        return ht[r] - ht[l - 1] * g[r - l + 1];
    }

    public static void main(String[] args) throws IOException {
        BufferedReader br = new BufferedReader(new InputStreamReader(System.in));
        PrintWriter out = new PrintWriter(System.out);

        g[0] = 1;
        for (int i = 1; i < N; i++) {
            g[i] = g[i - 1] * M;
        }

        String sStr = br.readLine();
        if (sStr == null) return;
        sStr = sStr.trim();
        int n = sStr.length();
        
        for (int i = 1; i <= n; i++) {
            s[i] = sStr.charAt(i - 1);
        }
        s[++n] = '?';

        int num = 0;
        for (int i = 1; i <= n; i++) {
            hs[i] = hs[i - 1] * M + s[i];
            if (s[i] == '*' || s[i] == '?') {
                p[++num] = i;
            }
        }

        String tCountStr = br.readLine();
        if (tCountStr == null) return;
        int T = Integer.parseInt(tCountStr.trim());

        while (T-- > 0) {
            String tStr = br.readLine();
            if (tStr == null) break;
            tStr = tStr.trim();
            int m = tStr.length();
            
            for (int i = 1; i <= m; i++) {
                t[i] = tStr.charAt(i - 1);
            }
            
            t[++m] = '$';

            for (int i = 1; i <= m; i++) {
                ht[i] = ht[i - 1] * M + t[i];
            }

            for (int i = 0; i <= num; i++) {
                for (int j = 0; j <= m; j++) {
                    f[i][j] = false;
                }
            }

            f[0][0] = true;

            for (int i = 0; i < num; i++) {
                if (s[p[i]] == '*') {
                    for (int j = 1; j <= m; j++) {
                        if (f[i][j - 1]) f[i][j] = true;
                    }
                }

                int len = p[i + 1] - p[i];
                for (int j = 0; j <= m - len + 1; j++) {
                    if (f[i][j] && hashs(p[i] + 1, p[i + 1] - 1) == hasht(j + 1, j + len - 1)) {
                        int offset = (s[p[i + 1]] != '?') ? 1 : 0;
                        f[i + 1][j + len - offset] = true;
                    }
                }
            }

            if (f[num][m]) {
                out.println("YES");
            } else {
                out.println("NO");
            }
        }
        
        out.flush();
        out.close();
        br.close();
    }
}
```

## G 汉诺塔

```python
import java.io.InputStream;
import java.io.IOException;

public class Main {
    static class FastScanner {
        private InputStream stream;
        private byte[] buf = new byte[1 << 16];
        private int head, tail;

        public FastScanner(InputStream stream) {
            this.stream = stream;
        }

        private int read() {
            if (head >= tail) {
                head = 0;
                try {
                    tail = stream.read(buf, 0, buf.length);
                } catch (IOException e) {
                    return -1;
                }
                if (tail <= 0) return -1;
            }
            return buf[head++];
        }

        public long nextLong() {
            int c = read();
            while (c <= 32) {
                if (c == -1) return -1;
                c = read();
            }
            boolean negative = false;
            if (c == '-') {
                negative = true;
                c = read();
            }
            long res = 0;
            while (c > 32) {
                res = res * 10 + c - '0';
                c = read();
            }
            return negative ? -res : res;
        }


        public char nextAlphaChar() {
            int c = read();
            while (c != -1 && !isAlpha(c)) {
                c = read();
            }
            return (char) c;
        }

        private boolean isAlpha(int c) {
            return (c >= 'A' && c <= 'Z') || (c >= 'a' && c <= 'z');
        }
    }

    public static void main(String[] args) {
        FastScanner in = new FastScanner(System.in);

        long nInput = in.nextLong();
        if (nInput == -1) return; 
        int n = (int) nInput;

        long[][] f = new long[36][4];
        int[][] g = new int[36][4];
        boolean[] vis = new boolean[4];

        for (int t = 1; t <= 6; t++) {
            char U = in.nextAlphaChar();
            char V = in.nextAlphaChar();
            int u = U - 'A' + 1;
            int v = V - 'A' + 1;
            
            if (vis[u]) continue;
            vis[u] = true;
            
            f[1][u] = 1;
            g[1][u] = v;
        }

        for (int i = 2; i <= n; i++) {
            for (int x = 1; x <= 3; x++) {
                int y = g[i - 1][x];
                int z = 6 - x - y;
                
                if (g[i - 1][y] == z) {
                    f[i][x] = f[i - 1][x] + 1 + f[i - 1][y];
                    g[i][x] = z;
                } else {
                    f[i][x] = f[i - 1][x] + 1 + f[i - 1][y] + 1 + f[i - 1][x];
                    g[i][x] = y;
                }
            }
        }

        System.out.println(f[n][1]);
    }
}
```

## H 马步距离

```python
## add your code here
import java.io.InputStream;
import java.io.IOException;
import java.util.Arrays;

public class Main {
    
    static class FastScanner {
        private InputStream stream;
        private byte[] buf = new byte[1024];
        private int head, tail;

        public FastScanner(InputStream stream) {
            this.stream = stream;
        }

        public int read() {
            if (head >= tail) {
                head = 0;
                try {
                    tail = stream.read(buf, 0, buf.length);
                } catch (IOException e) {
                    return -1;
                }
                if (tail <= 0) return -1;
            }
            return buf[head++];
        }

        public int nextInt() {
            int c = read();
            while (c <= 32) {
                if (c == -1) return -1;
                c = read();
            }
            boolean negative = false;
            if (c == '-') {
                negative = true;
                c = read();
            }
            int res = 0;
            while (c > 32) {
                res = res * 10 + c - '0';
                c = read();
            }
            return negative ? -res : res;
        }
    }

    static final int[] dx = {-2, -2, -1, -1, 1, 1, 2, 2};
    static final int[] dy = {-1, 1, -2, 2, -2, 2, -1, 1};

    public static void main(String[] args) {
        FastScanner sc = new FastScanner(System.in);
        
        int ox = sc.nextInt();
        if (ox == -1)
            return;
        int oy = sc.nextInt();
        int ex = sc.nextInt();
        int ey = sc.nextInt();

        int ans = 0;
        int x = Math.abs(ox - ex);
        int y = Math.abs(oy - ey);

        while (x + y > 50) {
            if (x < y) {
                int temp = x;
                x = y;
                y = temp;
            }
            if (x - 4 > (y << 1)) {
                x -= 4;
            } else {
                x -= 4;
                y -= 2;
            }
            ans += 2;
        }
        
        x += 50;
        y += 50;

        // BFS 初始化
        int[][] dis = new int[105][105];
        for (int i = 0; i < 105; i++) {
            Arrays.fill(dis[i], -1);
        }

        int[] qx = new int[10005];
        int[] qy = new int[10005];
        int head = 0, tail = 0;

        qx[tail] = x;
        qy[tail] = y;
        tail++;
        dis[x][y] = 0;
        
        while (head < tail) {
            int Nx = qx[head];
            int Ny = qy[head];
            head++;
            if (Nx == 50 && Ny == 50) {
                break;
            }
            for (int k = 0; k < 8; k++) {
                int tx = Nx + dx[k];
                int ty = Ny + dy[k];
                if (tx > 0 && ty > 0 && tx <= 100 && ty <= 100 && dis[tx][ty] == -1) {
                    dis[tx][ty] = dis[Nx][Ny] + 1;
                    qx[tail] = tx;
                    qy[tail] = ty;
                    tail++;
                }
            }
        }
        System.out.println(ans + dis[50][50]);
    }
}
```

## I 直方图最大矩形

```python
## add your code here
import java.util.*;

public class Solution {
    public int largestRectangleArea(int[] heights) {
        int n = heights.length;
        Deque<Integer> st = new ArrayDeque<>();
        st.push(-1);
        int ans = 0;
        for (int right = 0; right <= n; right++) {
            int h = right < n ? heights[right] : -1;
            while (st.size() > 1 && heights[st.peek()] >= h) {
                int i = st.pop();
                int left = st.peek();
                ans = Math.max(ans, heights[i] * (right - left - 1));
            }
            st.push(right);
        }
        return ans;

    }
}
```

```python
## add your code here
import java.io.InputStream;
import java.io.IOException;

public class Main {
    static class FastReader {
        private InputStream stream;
        private byte[] buf = new byte[1024];
        private int curChar;
        private int numChars;

        public FastReader(InputStream stream) {
            this.stream = stream;
        }

        public int read() {
            if (numChars == -1) throw new RuntimeException();
            if (curChar >= numChars) {
                curChar = 0;
                try {
                    numChars = stream.read(buf, 0, 1024);
                } catch (IOException e) {
                    throw new RuntimeException();
                }
                if (numChars <= 0) return -1;
            }
            return buf[curChar++];
        }

        public int nextInt() {
            int c = read();
            while (isSpaceChar(c)) c = read();
            int res = 0;
            do {
                res = res * 10 + c - '0';
                c = read();
            } while (!isSpaceChar(c));
            return res;
        }

        private boolean isSpaceChar(int c) {
            return c == ' ' || c == '\n' || c == '\r' || c == '\t' || c == -1;
        }
    }

    public static void main(String[] args) {
        FastReader in = new FastReader(System.in);
        
        int n = in.nextInt();
        if (n <= 0) return;

        int[] head = new int[n + 1];
        int[] next = new int[n + 1];
        int[] to = new int[n + 1];
        int edgeCount = 0;

        for (int i = 2; i <= n; i++) {
            int x = in.nextInt();
            edgeCount++;
            to[edgeCount] = i;
            next[edgeCount] = head[x];
            head[x] = edgeCount;
        }

        int[][] f = new int[n + 1][5];

        for (int i = n; i >= 1; i--) {
            f[i][0] = 1;
            int t1 = 10000;
            int t2 = 10000;

            for (int e = head[i]; e != 0; e = next[e]) {
                int p = to[e];
                
                f[i][3] += f[p][2];
                f[i][4] += f[p][3];
                f[i][0] += f[p][4];
                
                t1 = Math.min(t1, f[p][0] - f[p][3]);
                t2 = Math.min(t2, f[p][1] - f[p][2]);
            }
            
            f[i][1] = f[i][4] + t1;
            f[i][2] = Math.min(f[i][3] + t2, Math.min(f[i][0], f[i][1]));
            f[i][3] = Math.min(f[i][2], f[i][3]);
            f[i][4] = Math.min(f[i][4], f[i][3]);
        }

        System.out.println(f[1][2]);
    }
}
```
