idea:

````python
'''
算法方案分析：
    为了在时间效率未O(logn)的情况下找到最小值，我们可以使用二分查找的方法来实现。首先我们需要找到数组的中间位置，
然后比较中间位置的元素与它的左右元素，如果中间位置的元素小于它的左右元素，那么说明这个元素就是最小值，直接返回即可。
如果中间位置的元素大于它的右边元素，那么说明右边有最小值，否则最小值在左边，因此就可以通过不断地缩小搜索范围来找到最小值。
'''

# add your code here
A=[9, 3, 7, 2, 1, 4, 5 ]
def binary_search_min(A, low, high):
    while low <= high:
        mid = (low + high) // 2
        if A[mid] < A[mid + 1] and A[mid] < A[mid - 1]:
            return A[mid]
        elif A[mid] < A[mid + 1]:
            high = mid
            return binary_search_min(A, low, high)
        else:
            low = mid
            return binary_search_min(A, low, high)

ans = binary_search_min(A, 0, len(A) - 1)
print(ans)

# your algorithm time complexity is: O(log n)

# 2、找出所有的最小值
ans = []
for i in range(1, len(A) - 1):
    if A[i] < A[i - 1] and A[i] < A[i + 1]:
        ans.append(A[i])

print(ans)
````

## 问题 7

> 0-1背包问题：现在有4块大理石，每个大理石对应的重量和价值使用一个元组表示, 即（重量，价值），4块大理石的重量和价值为：[(5,10), (4,40), (6,30), (3, 50)], 假设你有一辆最大承重为9的小推车，请问使用这个小推车装大理石的最大价值为多少

请设计一个算法解决该问题，分析算法设计思路，计算时间复杂度，并基于python编程实现

idea:

````python
'''
算法方案分析：
    这是一个经典的背包问题，可以通过动态规划进行求解，我们可以通过一个二维的数组来记录前i个物品中，
背包容量为w时的最大价值，当第i个物品的重量超过背包容量w时，不能放入背包，所以此时的最大价值等于前i-1
个物品在容量为w时的最大价值；而当第i个物品可以放入时，只需选择放入或不放入第i个物品中价值较大的情况。
因此，可以通过两层for循环来实现。
'''

# add your code here
W = 9
N = 4
dp = [[0 for _ in range(W + 1)] for _ in range(N + 1)]
weight = [5, 4, 6, 3]
value = [10, 40, 30, 50]
for i in range(1,N + 1):
    for w in range(1, W + 1):
        # 第i个物品重量超过背包容量w,不能放入背包
        if w - weight[i - 1] < 0:
            dp[i][w] = dp[i-1][w]
        else:
            dp[i][w] = max(dp[i -1][w], dp[i-1][w - weight[i-1]] + value[i -1])

print(dp[N][W])

# your algorithm time complexity is: O(N*W)
````

idea:

````python
'''
算法方案分析：
    首先，我们需要对二叉搜索树进行中序遍历，得到一个升序的数组，然后我们需要找到两个元素的位置，
这两个元素满足前一个元素大于后一个元素，如果只找到一个元素，那么说明这两个元素是相邻的，如果找到
两个元素，那么说明这两个元素是不相邻的，因此，我们需要记录下这两个元素的位置，然后交换这两个元素
的位置即可在不破坏原结构的情况下恢复二叉搜索树。
'''

# add your code here

def swap(arr, i, j):
    temp = arr[i]
    arr[i] = arr[j]
    arr[j] = temp 

# 二叉树的中序遍历得到升序的数组，此处我们只用记录下索引的位置即可

def inorder_traverse(arr, root):
    
    if root >= len(arr) or arr[root] == None:
        return []

    left, right = 2 * root + 1, 2 * root + 2
    
    return inorder_traverse(arr, left) + [root] + inorder_traverse(arr, right)
    

def maintain_bst(arr):
    if not arr:
        return arr
    
    indices = inorder_traverse(arr, 0)

    first_index, second_index = None, None
    
    for i in range(len(indices) - 1):
        current_index = indices[i]
        next_index = indices[i + 1]

        if arr[current_index] > arr[next_index]:
            if first_index == None:
                first_index = current_index
                second_index = next_index
            else:
                second_index = next_index

    if first_index != None and second_index != None:
        swap(arr, first_index, second_index)

    return arr


arr = [1, 3, None, None, 2]
arr1 = [3, 1, 4, None, None, 2]
arr2 = [7, 5, 9, 1, 4, None, 10]

print(maintain_bst(arr))
print(maintain_bst(arr1))
print(maintain_bst(arr2))



# your algorithm time complexity is: 
````

解释：从根到叶子节点路径 4->9->5 代表数字 495，4->9->1 代表数字 491，4->0 代表数字 40。因此和为 495 + 491 + 40 = 1026 。

idea:

````python
'''
算法方案分析：
    首先，需要对二叉树进行深度优先搜索，计算从根节点到每个叶子节点的路径和。
    然后，判断当前节点是否为叶子节点，如果是叶子节点，那么我们就将当前路径和加入到结果中，
如果不是叶子节点，那么我们就继续递归搜索左右子树，直到找到所有的叶子节点为止。
'''

# add your code here

def DFS(arr, root, current_path):
    if root >= len(arr) or arr[root] is None:
        return 0
    current_path = current_path * 10 + arr[root]
    left = 2 * root + 1
    right = 2 * root + 2

    is_left_empty = (left >= len(arr) or arr[left] is None)
    is_right_empty = (right >= len(arr) or arr[right] is None)

    if is_left_empty and is_right_empty:
        return current_path

    return  DFS(arr, left, current_path) + DFS(arr, right, current_path)
    

root = [4,9,0,5,1]
sum = 0   
res = DFS(root, 0, sum)
print(res)       

# your algorithm time complexity is: O(n)
````
